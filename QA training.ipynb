```python
import numpy as np # linear algebra
import pandas as pd # data processing, CSV file I/O (e.g. pd.read_csv)

from tqdm.notebook import tqdm
tqdm.pandas(desc="Encoding Questions")


from sentence_transformers import SentenceTransformer, util
```

```python
#Load data
df = pd.read_csv(r'C:\Users\ABHISHEK\project\chatbot final ques dataset.csv')
```

```python
df
```

```python
#dropping first column since its not needed
df=df.drop("Unnamed: 3",axis=1)
```

```python
#loading model
model = SentenceTransformer('paraphrase-MiniLM-L6-v2')
```

```python
df = df.dropna(subset=['User Question'])
df['question_embedding'] = df['User Question'].progress_apply(
    lambda x: model.encode(str(x), convert_to_tensor=True)
)
```

```python
df
```

```python
df["Answer"]=df["AI Response"]
```

```python
# Encode all questions in the dataset
df['question_embedding'] = df['User Question'].progress_apply(lambda x: model.encode(x, convert_to_tensor=True))
```

```python
def find_most_similar_answer(user_question):

    '''
    It retrieves the answer to the most similar question from a dataset based on
    user's query using a Sentence Transformer model to compute similarities.

    '''
    # Encode user's question
    user_question_embedding = model.encode(user_question, convert_to_tensor=True)

    # Compute similarities
    df['similarity'] = df['question_embedding'].apply(lambda x: util.pytorch_cos_sim(user_question_embedding, x).item())

    # Find the most similar question
    most_similar_row = df.loc[df['similarity'].idxmax()]

    return most_similar_row['User Question'], most_similar_row['Answer']
```

```python
#Example use-case

user_question = "what is flu"
question, answer = find_most_similar_answer(user_question)
print(f"Similar question: {question}\n\nAnswer: {answer}")
```
